```python
# ==========================================
# DATASET SPLITTING SCRIPT (TRAIN + VAL)
# ==========================================

import os
import shutil
import random

# -----------------------------
# USER SETTINGS
# -----------------------------
ORIGINAL_DATASET_DIR = "F:/Image"   # contains open/ and closed/
OUTPUT_DATASET_DIR   = "F:/Split"            # new dataset folder
TRAIN_RATIO = 0.8                            # 80% train, 20% validation
SEED = 42                                   # for reproducibility

random.seed(SEED)

# -----------------------------
# CREATE REQUIRED FOLDERS
# -----------------------------
folders = [
    "train/open",
    "train/closed",
    "val/open",
    "val/closed"
]

for folder in folders:
    os.makedirs(os.path.join(OUTPUT_DATASET_DIR, folder), exist_ok=True)

# -----------------------------
# FUNCTION TO SPLIT DATA
# -----------------------------
def split_class(class_name):
    src_dir = os.path.join(ORIGINAL_DATASET_DIR, class_name)
    images = os.listdir(src_dir)

    # Shuffle images
    random.shuffle(images)

    # Calculate split index
    split_index = int(len(images) * TRAIN_RATIO)

    train_images = images[:split_index]
    val_images   = images[split_index:]

    # Copy training images
    for img in train_images:
        src_path = os.path.join(src_dir, img)
        dst_path = os.path.join(OUTPUT_DATASET_DIR, "train", class_name, img)
        shutil.copy2(src_path, dst_path)

    # Copy validation images
    for img in val_images:
        src_path = os.path.join(src_dir, img)
        dst_path = os.path.join(OUTPUT_DATASET_DIR, "val", class_name, img)
        shutil.copy2(src_path, dst_path)

    # Print summary
    print(f"{class_name.upper()} class:")
    print(f"  Total images     : {len(images)}")
    print(f"  Training images  : {len(train_images)}")
    print(f"  Validation images: {len(val_images)}\n")

# -----------------------------
# RUN SPLITTING
# -----------------------------
split_class("open")
split_class("closed")

print("✅ Dataset splitting completed successfully!")
```
