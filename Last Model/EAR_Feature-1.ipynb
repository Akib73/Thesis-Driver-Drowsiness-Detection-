```python
# =========================================================
# IMPORTS
# =========================================================
import os
import cv2
import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
from sklearn.model_selection import KFold
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    confusion_matrix, roc_curve, auc, roc_auc_score
)
import matplotlib.pyplot as plt
from PIL import Image

# EAR
import dlib
from scipy.spatial import distance as dist

# =========================================================
# DEVICE
# =========================================================
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# =========================================================
# SEPARATE TRANSFORMS (Train vs Test) ✅
# =========================================================
train_transform = transforms.Compose([
    transforms.Resize((64, 64)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(15),
    transforms.RandomAffine(degrees=10, translate=(0.1, 0.1)),
    transforms.ColorJitter(brightness=0.3, contrast=0.3),
    transforms.GaussianBlur(3),
    transforms.ToTensor(),
    transforms.Normalize([0.5], [0.5])
])

# Test transform: NO RANDOM AUGMENTATION
test_transform = transforms.Compose([
    transforms.Resize((64, 64)),
    transforms.ToTensor(),
    transforms.Normalize([0.5], [0.5])
])

# =========================================================
# DATASET (Accepts transform and indices)
# =========================================================
class EyeDataset(Dataset):
    def __init__(self, root_dir, transform=None, indices=None):
        """
        Args:
            root_dir (str): Path to the dataset root containing 'open' and 'close' folders.
            transform (callable, optional): Transform to apply to images.
            indices (list, optional): List of indices to use; if None, use all.
        """
        self.transform = transform
        self.images, self.labels = [], []

        for label, cls in enumerate(["open", "close"]):
            path = os.path.join(root_dir, cls)
            for img in os.listdir(path):
                if img.lower().endswith((".jpg", ".png", ".jpeg")):
                    self.images.append(os.path.join(path, img))
                    self.labels.append(label)

        # If indices are given, filter the data
        if indices is not None:
            self.images = [self.images[i] for i in indices]
            self.labels = [self.labels[i] for i in indices]

    def __len__(self):
        return len(self.images)

    def __getitem__(self, idx):
        # Load as grayscale
        img = cv2.imread(self.images[idx], 0)
        # Improve contrast (helps with glasses/blur)
        img = cv2.equalizeHist(img)
        # Convert to PIL for transforms
        img = Image.fromarray(img)

        if self.transform:
            img = self.transform(img)

        label = self.labels[idx]
        return img, label


# =========================================================
# IMPROVED CNN (RESIDUAL STYLE)
# =========================================================
class ResidualBlock(nn.Module):
    def __init__(self, in_ch):
        super().__init__()
        self.conv1 = nn.Conv2d(in_ch, in_ch, 3, padding=1)
        self.bn1 = nn.BatchNorm2d(in_ch)
        self.conv2 = nn.Conv2d(in_ch, in_ch, 3, padding=1)
        self.bn2 = nn.BatchNorm2d(in_ch)

    def forward(self, x):
        identity = x
        out = torch.relu(self.bn1(self.conv1(x)))
        out = self.bn2(self.conv2(out))
        out += identity
        return torch.relu(out)


class ImprovedCNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.layer1 = nn.Sequential(
            nn.Conv2d(1, 16, 3, padding=1),
            nn.BatchNorm2d(16),
            nn.ReLU(),
            nn.MaxPool2d(2)
        )
        self.layer2 = nn.Sequential(
            nn.Conv2d(16, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.MaxPool2d(2)
        )
        self.res = ResidualBlock(32)
        self.layer3 = nn.Sequential(
            nn.Conv2d(32, 64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.MaxPool2d(2)
        )
        self.fc = nn.Sequential(
            nn.Flatten(),
            nn.Linear(64 * 8 * 8, 128),
            nn.ReLU(),
            nn.Dropout(0.5),
            nn.Linear(128, 2)
        )

    def forward(self, x):
        x = self.layer1(x)
        x = self.layer2(x)
        x = self.res(x)
        x = self.layer3(x)
        x = self.fc(x)
        return x


# =========================================================
# EAR FUNCTIONS
# =========================================================
detector = dlib.get_frontal_face_detector()
predictor = dlib.shape_predictor("shape_predictor_68_face_landmarks.dat")


def ear_calc(eye):
    A = dist.euclidean(eye[1], eye[5])
    B = dist.euclidean(eye[2], eye[4])
    C = dist.euclidean(eye[0], eye[3])
    return (A + B) / (2.0 * C)


def compute_ear(img):
    """Takes a PIL image or numpy array, returns EAR or None."""
    if isinstance(img, Image.Image):
        img = np.array(img)
    
    faces = detector(img, 0)
    if len(faces) == 0:
        return None
    
    shape = predictor(img, faces[0])
    pts = np.array([[p.x, p.y] for p in shape.parts()])
    left = ear_calc(pts[42:48])
    right = ear_calc(pts[36:42])
    return (left + right) / 2.0


def compute_average_ear(dataset, indices):
    """Helper to compute mean EAR over specific indices (Training set)."""
    ears = []
    for idx in indices:
        img, _ = dataset[idx]  # dataset is the original (no transform) to get PIL
        ear = compute_ear(img)
        if ear is not None:
            ears.append(ear)
    if ears:
        return np.mean(ears)
    return None


# =========================================================
# MAIN LOOP
# =========================================================
if __name__ == "__main__":
    # Create output directory for models
    os.makedirs("saved_models", exist_ok=True)

    # Create a base dataset (without transform) to use for splitting and EAR computation
    base_dataset = EyeDataset("C:/Users/ASUS/Desktop/Image", transform=None)
    kf = KFold(n_splits=5, shuffle=True, random_state=42)

    # Lists to store overall metrics
    all_true, all_score = [], []
    fold_acc, fold_prec, fold_rec, fold_f1, fold_auc = [], [], [], [], []

    for fold, (train_idx, test_idx) in enumerate(kf.split(base_dataset)):

        print(f"\n{'='*20} FOLD {fold+1} {'='*20}")

        # --- Create train and test datasets with appropriate transforms ---
        train_set = EyeDataset(
            "C:/Users/ASUS/Desktop/Image",
            transform=train_transform,
            indices=train_idx
        )
        test_set = EyeDataset(
            "C:/Users/ASUS/Desktop/Image",
            transform=test_transform,
            indices=test_idx
        )

        train_loader = DataLoader(train_set, batch_size=32, shuffle=True)
        test_loader = DataLoader(test_set, batch_size=32, shuffle=False)

        # --- Model ---
        model = ImprovedCNN().to(device)
        opt = torch.optim.Adam(model.parameters(), lr=0.001)
        loss_fn = nn.CrossEntropyLoss()

        # --- TRAINING WITH LOSS AND ACCURACY RECORDING (Train + Validation) ---
        train_losses, train_accs = [], []
        val_losses, val_accs = [], []

        for epoch in range(10):
            # ---------- Training ----------
            model.train()
            total, correct, loss_sum = 0, 0, 0.0

            for x, y in train_loader:
                x, y = x.to(device), y.to(device)

                opt.zero_grad()
                out = model(x)
                loss = loss_fn(out, y)
                loss.backward()
                opt.step()

                loss_sum += loss.item() * x.size(0)  # sum loss over batch
                pred = out.argmax(1)
                correct += (pred == y).sum().item()
                total += y.size(0)

            train_acc = correct / total
            train_loss = loss_sum / total
            train_accs.append(train_acc)
            train_losses.append(train_loss)

            # ---------- Validation (using test set) ----------
            model.eval()
            val_correct, val_total = 0, 0
            val_loss_sum = 0.0
            with torch.no_grad():
                for x_val, y_val in test_loader:
                    x_val, y_val = x_val.to(device), y_val.to(device)
                    out_val = model(x_val)
                    loss_val = loss_fn(out_val, y_val)
                    val_loss_sum += loss_val.item() * x_val.size(0)
                    pred_val = out_val.argmax(1)
                    val_correct += (pred_val == y_val).sum().item()
                    val_total += y_val.size(0)

            val_acc = val_correct / val_total
            val_loss = val_loss_sum / val_total
            val_accs.append(val_acc)
            val_losses.append(val_loss)

            print(f"Epoch {epoch+1} | Train Acc: {train_acc:.4f} | Train Loss: {train_loss:.4f} | Val Acc: {val_acc:.4f} | Val Loss: {val_loss:.4f}")

        # --- Plot Training & Validation Curves (Loss + Accuracy) ---
        epochs_range = range(1, 11)
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
        fig.suptitle(f'Fold {fold+1} - Training and Validation Curves')

        # Loss plot
        ax1.plot(epochs_range, train_losses, marker='o', label='Train Loss', color='blue')
        ax1.plot(epochs_range, val_losses, marker='s', label='Validation Loss', color='red')
        ax1.set_xlabel('Epoch')
        ax1.set_ylabel('Loss')
        ax1.legend()
        ax1.grid(alpha=0.3)

        # Accuracy plot
        ax2.plot(epochs_range, train_accs, marker='o', label='Train Accuracy', color='blue')
        ax2.plot(epochs_range, val_accs, marker='s', label='Validation Accuracy', color='red')
        ax2.set_xlabel('Epoch')
        ax2.set_ylabel('Accuracy')
        ax2.legend()
        ax2.grid(alpha=0.3)

        plt.tight_layout()
        plt.show()

        # --- Save Model ---
        model_path = f"saved_models/model_fold_{fold+1}.pth"
        torch.save(model.state_dict(), model_path)
        print(f"Model saved to {model_path}")

        # --- DYNAMIC EAR THRESHOLDS (Calculated ONLY on Training set) ---
        mean_ear = compute_average_ear(base_dataset, train_idx)
        if mean_ear is not None:
            LOW = mean_ear * 0.70   # 30% below average = definitely closed
            HIGH = mean_ear * 0.88  # 12% below average = definitely open
            print(f"Calibrated EAR: Mean={mean_ear:.3f} | LOW={LOW:.3f} | HIGH={HIGH:.3f}")
        else:
            # Fallback if dlib fails on all training images
            LOW, HIGH = 0.2, 0.3
            print("Warning: No faces in training set. Using static thresholds (0.2, 0.3).")

        # --- EVALUATION (EAR FUSION ON LOGITS) ---
        model.eval()
        y_true, y_pred, y_score = [], [], []

        with torch.no_grad():
            for x, y in test_loader:
                x, y = x.to(device), y.to(device)

                # Get RAW LOGITS (NOT softmax)
                logits = model(x)  # Shape: [batch, 2]

                for i in range(len(x)):
                    # 1. CNN Logit for "Closed" (class 1)
                    base_logit = logits[i][1].item()

                    # 2. Convert tensor back to PIL for EAR
                    img_tensor = x[i].cpu()
                    # Denormalize from [-1, 1] to [0, 255]
                    img_np = ((img_tensor.numpy().squeeze() * 0.5 + 0.5) * 255).astype(np.uint8)
                    img_pil = Image.fromarray(img_np)

                    ear = compute_ear(img_pil)

                    # 3. EAR Adjustment (Added to LOGIT, NOT probability)
                    if ear is not None:
                        if ear < LOW:      # Eyes closed
                            ear_adj = 1.5
                        elif ear > HIGH:   # Eyes open
                            ear_adj = -1.5
                        else:
                            ear_adj = 0.0
                    else:
                        # If dlib fails, we trust the CNN completely
                        ear_adj = 0.0

                    # 4. Mathematically correct fusion
                    fused_logit = base_logit + ear_adj

                    # 5. Convert to probability via Sigmoid
                    final_prob = 1 / (1 + np.exp(-fused_logit))
                    pred = 1 if final_prob > 0.5 else 0

                    y_true.append(y[i].item())
                    y_pred.append(pred)
                    y_score.append(final_prob)

        # --- PER-FOLD METRICS ---
        acc = accuracy_score(y_true, y_pred)
        prec = precision_score(y_true, y_pred, average='binary')
        rec = recall_score(y_true, y_pred, average='binary')
        f1 = f1_score(y_true, y_pred, average='binary')
        auc_score = roc_auc_score(y_true, y_score)

        fold_acc.append(acc)
        fold_prec.append(prec)
        fold_rec.append(rec)
        fold_f1.append(f1)
        fold_auc.append(auc_score)

        print(f"\n--- Fold {fold+1} Results ---")
        print(f"Accuracy:  {acc:.4f}")
        print(f"Precision: {prec:.4f}")
        print(f"Recall:    {rec:.4f}")
        print(f"F1-Score:  {f1:.4f}")
        print(f"AUC:       {auc_score:.4f}")

        # --- Confusion Matrix Plot ---
        cm = confusion_matrix(y_true, y_pred)
        plt.figure(figsize=(5, 4))
        plt.imshow(cm, cmap='Blues', interpolation='nearest')
        plt.title(f"Confusion Matrix - Fold {fold+1}")
        plt.colorbar()
        plt.xlabel("Predicted")
        plt.ylabel("True")
        plt.xticks([0, 1], ["Open", "Close"])
        plt.yticks([0, 1], ["Open", "Close"])
        # Add text annotations
        for i in range(2):
            for j in range(2):
                plt.text(j, i, str(cm[i, j]), ha='center', va='center', color='red')
        plt.show()

        # Store for global ROC
        all_true.extend(y_true)
        all_score.extend(y_score)

    # =========================================================
    # FINAL OVERALL METRICS
    # =========================================================
    print("\n" + "="*50)
    print("FINAL RESULTS (Across 5 Folds)")
    print("="*50)
    print(f"Average Accuracy:  {np.mean(fold_acc):.4f} (+/- {np.std(fold_acc):.4f})")
    print(f"Average Precision: {np.mean(fold_prec):.4f}")
    print(f"Average Recall:    {np.mean(fold_rec):.4f}")
    print(f"Average F1-Score:  {np.mean(fold_f1):.4f}")
    print(f"Average AUC:       {np.mean(fold_auc):.4f}")

    # --- Global ROC Curve ---
    fpr, tpr, _ = roc_curve(all_true, all_score)
    global_auc = auc(fpr, tpr)

    plt.figure(figsize=(7, 5))
    plt.plot(fpr, tpr, linewidth=2, label=f"ROC Curve (AUC = {global_auc:.3f})")
    plt.plot([0, 1], [0, 1], 'k--', linewidth=1)
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.title("Global ROC Curve")
    plt.legend(loc="lower right")
    plt.grid(alpha=0.3)
    plt.show()

    print(f"\nGlobal AUC: {global_auc:.4f}")
    print("All models saved in './saved_models/' for further testing.")
```
